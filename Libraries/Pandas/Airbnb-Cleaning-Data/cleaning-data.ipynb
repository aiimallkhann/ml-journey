DATA CLEANING:

# Importing The Dataset

```python
import pandas as pd
from pathlib import Path

BASE = Path.cwd()
data = pd.read_csv(BASE / 'archive (1)' / 'Airbnb_Open_Data.csv')

#^CHECK ALL THE COLUMNS:
(data.columns)
```

# Creating Two Columns For Keeping And Dropping

```python
#& delete the redundant columns:
columnsToKeep = ['NAME', 'host id', 'host_identity_verified', 'host name',
       'neighbourhood group', 'neighbourhood', 'lat', 'long', 'country',
       'country code', 'instant_bookable', 'cancellation_policy', 'room type',
       'Construction year', 'price', 'service fee', 'minimum nights',
       'number of reviews', 'last review']
columnsToDrop = ['id','reviews per month',
       'review rate number', 'calculated host listings count',
       'availability 365', 'house_rules', 'license']
```

```python
len(columnsToKeep)
```

```python
len(columnsToDrop)
```

# 1st Method, Filtering Data

```python
df = data[columnsToKeep]
```

# 2nd Method, Dropping Columns

```python
data.drop(columns=columnsToDrop, inplace=True)
```

# Renaming The Columns

```python
data.rename(columns={'NAME' : 'Name'}, inplace=True)
```

# 2nd Method Of Renaming Columns

```python
newColumnNames = []
for i in data.columns:
    newColumnNames.append(i.upper())
```

```python
newColumnNames
```

```python
#This copies the already modified (19 columns) data frame.
newDataframeToModify = data.copy()
#This copies the upper case names from newColumnNames
newDataframeToModify.columns = newColumnNames
#This prints the new Dataframe with upper case letters
newDataframeToModify.columns
```

# Data Duplication Techniques

```python
#Prints whether the first 5 of the series are duplicated
data.duplicated().head()
```

```python
#Prints whether the last 5 of the series are duplicated
data.duplicated().tail()
```

```python
#This checks how many total are duplicated and how many aren't
data.duplicated().value_counts()
```

```python
#This will create the data frame with no duplicates and will make a new dataframe.
data.drop_duplicates(inplace=True)
```

```python
#0 in the output shows that the dataframe has no duplicates left.
data.duplicated().sum()
```

# Remove The NAN Values From The Dataset

```python
data.isna().sum()
```

# Drop All The NAN Values And Modify The DataFrame

```python
data.dropna(inplace=True)
data.isna().sum()
```

# Cleaning Individual Columns

```python
#This will make all the entries in the column uppercase
data['host_identity_verified'].str.upper()
```

```python
data['host_identity_verified'] = data['host_identity_verified'].str.upper()
data
```

# Turning False And True Into Numerical Values (0 = False, 1 = True)

```python
data['instant_bookable'].apply(lambda x: 1 if x==True else 0)
```

```python
data['instant_bookable'] = data['instant_bookable'].apply(lambda x: 1 if x==True else 0)
data
```

 2nd Method For Boolean Conversion To Int

```python
data['instant_bookable'] = data['instant_bookable'].map({True:1, False: 0}).astype('Int64')
```

# Reseting Index

```python
data.reset_index(inplace=True)
```

```python
check = 0
for i in data['instant_bookable']:
    check += (data['instant_bookable'] == 1).sum()
```

```python
check
```
